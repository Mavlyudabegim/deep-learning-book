```python
import numpy as np
import matplotlib.pyplot as plt
 # random variable
x = np.random.uniform(low=0, high=1, size=10)  # 10 samples from a uniform distribution
x
```

```python
# marginal probability distribution

# height of students in a class
X = [150, 160, 170]

# health status of students in a class
Y = [1, 0] # 1 = healthy, 0 = sick

p = [[0.15, 0.05], [0.3, 0.1], [0.25, 0.15]] # joint probability distribution

# marginal probability distribution of X=150

p_X = np.sum(p[0])
print(p_X)

# joint probability distribution of X=150 and Y=1
p_XY = p[0][0]
print(p_XY)

# conditional probability distribution of Y given X=150
p_Y_given_X = p_XY / p_X
print(p_Y_given_X)
```

```python
# Gaussian distribution
mean = 0
std_dev = 1
samples = np.random.normal(loc=mean, scale=std_dev, size=10)
samples
```

```python
x = np.linspace(-6, 6, 1000)
pdf = (1 / np.sqrt(2 * np.pi * std_dev**2)) * np.exp(-0.5 * ((x - mean) / std_dev)**2)

plt.plot(x, pdf)

plt.xlabel("x")
plt.ylabel("Density")
plt.title("Gaussian Distribution")
plt.show()
```

```python
# multivariate Gaussian distribution
mean = [0, 0]
cov = [[1, 0], [0, 1]]  # diagonal covariance matrix
samples = np.random.multivariate_normal(mean, cov, size=10)
samples
```

```python
# Laplace distribution
mean = 0
scale = 1
samples = np.random.laplace(loc=mean, scale=scale, size=1000)
samples
```

```python
import matplotlib.pyplot as plt
# Plot histogram
plt.hist(samples, bins=50, density=True, alpha=0.6)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Histogram of Samples')
plt.show()
```

```python
# Theoretical Laplace PDF
x = np.linspace(-6, 6, 1000)
pdf = (1 / (2 * scale)) * np.exp(-np.abs(x - mean) / scale)

plt.plot(x, pdf)

plt.xlabel("x")
plt.ylabel("Density")
plt.title("Laplace Distribution")
plt.show()
```

```python
# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x)) 

x = np.linspace(-10, 10, 100)
y = sigmoid(x)
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Sigmoid Function")
plt.grid()
```

```python
# softplus function
def softplus(x):
    return np.log(1 + np.exp(x))

x = np.linspace(-10, 10, 100)
y = softplus(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("softplus(x)")
plt.title("Softplus Function")
plt.grid()
```

```python
# Bayes Rule
A = 0.01  # Probability of having a disease
B = 0.9  # Probability of testing positive P(B|A) = 0.9

not_A = 0.99  # Probability of not having the disease
P_B_not_A = 0.02  # Probability of testing positive given that you don't have the disease

#P(B)=P(B|A) * P(A) + P(B|not A) * P(not A)
P_B = B * A + P_B_not_A * not_A
P_A_given_B = (B * A) / P_B
print("Probability of having the disease given a positive test result: ", P_A_given_B)
```

```python
# KL divergence
def kl_divergence(p, q):
    return np.sum(p * np.log(p / q))    

p = np.array([0.1, 0.9])
q = np.array([0.2, 0.8])

d = kl_divergence(p, q)
print("KL Divergence: ", d)



```
